```python
category_sales = (
    df.groupby("category")["units_sold"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
category_returns = (
    df.groupby("category")["return_rate"]
    .mean()
    .sort_values(ascending=False)
)

category_returns
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_returns.values,
    y=category_returns.index
)

plt.title("Average Return Rate by Category")
plt.xlabel("Average Return Rate")
plt.ylabel("Category")

plt.show()
```

```python
outlier_summary = []

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
outlier_summary["Outlier_Percentage"] = (
    outlier_summary["Outlier_Count"] / len(df) * 100
).round(2)

outlier_summary
```

```python
category_returns = (
    df.groupby("category")["return_rate"]
    .mean()
    .sort_values(ascending=False)
)

print("Categories by average return rate:")
print(category_returns)
```

```python
high_rated_products = (
    df[df["review_count"] >= 20]
    .sort_values(
        ["rating", "review_count"],
        ascending=[False, False]
    )
)

high_rated_products[
    [
        "product_id",
        "category",
        "subcategory",
        "rating",
        "review_count",
        "units_sold"
    ]
].head(10)
```

```python
discount_sales = (
    df.groupby("discount_percent")["units_sold"]
    .agg(["mean", "median", "sum", "count"])
    .sort_index()
)

discount_sales




plt.figure(figsize=(10, 6))

sns.lineplot(
    data=discount_sales,
    x=discount_sales.index,
    y="mean",
    marker="o"
)

plt.title("Average Units Sold by Discount Percentage")
plt.xlabel("Discount (%)")
plt.ylabel("Average Units Sold")

plt.show()
```

```python
top_sales_category = category_sales.idxmax()
top_sales_value = category_sales.max()

highest_price_category = category_price.idxmax()
highest_price_value = category_price.max()

highest_return_category = category_returns.idxmax()
highest_return_value = category_returns.max()

highest_discount_category = category_discount.idxmax()
highest_discount_value = category_discount.max()

print("KEY BUSINESS INSIGHTS")
print("----------------------")

print(
    f"Highest total unit sales: {top_sales_category} "
    f"({top_sales_value:,.0f} units)"
)

print(
    f"Highest average price: {highest_price_category} "
    f"(${highest_price_value:.2f})"
)

print(
    f"Highest average return rate: {highest_return_category} "
    f"({highest_return_value:.2%})"
)

print(
    f"Highest average discount: {highest_discount_category} "
    f"({highest_discount_value:.2f}%)"
)
```

# Business Insights

## 1. Sales Performance

The category with the highest total number of units sold was identified
using category-level aggregation.

## 2. Pricing

Average product prices were compared across categories to identify
higher-priced product segments.

## 3. Returns

Average return rates were analyzed by category to identify categories
with relatively higher return behavior.

## 4. Discounts

Average discount percentages were compared across categories.
The relationship between discount levels and average units sold was
also investigated.

## 5. Customer/Product Ratings

Product ratings were analyzed together with review counts and sales
to understand whether highly rated products also show strong sales
performance.

## 6. Outliers

Statistical outliers were identified using the IQR method. Outliers
were investigated rather than automatically removed because unusual
values may represent legitimate premium products, bestsellers, or
other meaningful business cases.
