# Task 3 — Data Visualization Dashboard

## Objective

The objective of this project is to create meaningful visualizations
from the cleaned women's clothing dataset.

The analysis includes:

- Bar charts
- Line charts
- Pie charts
- Histograms
- Scatter plots

The visualizations are designed to communicate business insights
clearly and support data-driven decision making.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/cleaned_dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (10, 6)
```

```python
category_counts = df["category"].value_counts()

print(category_counts)
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)

plt.title("Number of Products by Category", fontsize=16)
plt.xlabel("Number of Products", fontsize=12)
plt.ylabel("Category", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)

plt.title("Number of Products by Category", fontsize=16)
plt.xlabel("Number of Products", fontsize=12)
plt.ylabel("Category", fontsize=12)

plt.tight_layout()

plt.savefig(
    "../outputs/products_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Insight

The bar chart shows the distribution of products across different
clothing categories. Categories with larger bars contain a greater
number of products in the dataset, while smaller bars represent
categories with fewer products.

```python
discount_sales = (
    df.groupby("discount_percent")["units_sold"]
    .mean()
    .reset_index()
)

discount_sales
```

```python
plt.figure(figsize=(12, 7))

sns.lineplot(
    data=discount_sales,
    x="discount_percent",
    y="units_sold",
    marker="o"
)

plt.title(
    "Average Units Sold by Discount Percentage",
    fontsize=16
)

plt.xlabel("Discount Percentage (%)", fontsize=12)
plt.ylabel("Average Units Sold", fontsize=12)

plt.xticks(discount_sales["discount_percent"])

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 7))

sns.lineplot(
    data=discount_sales,
    x="discount_percent",
    y="units_sold",
    marker="o"
)

plt.title(
    "Average Units Sold by Discount Percentage",
    fontsize=16
)

plt.xlabel("Discount Percentage (%)", fontsize=12)
plt.ylabel("Average Units Sold", fontsize=12)

plt.xticks(discount_sales["discount_percent"])

plt.tight_layout()

plt.savefig(
    "../outputs/average_sales_by_discount.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Insight

The line chart shows how average units sold vary across different
discount percentages. The pattern can be used to identify discount
levels associated with relatively higher or lower average sales.

The visualization shows an association between discount level and
sales; it does not establish that discounts directly cause changes
in sales.

```python
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=df,
    x="price_usd",
    y="units_sold",
    alpha=0.4
)

plt.title(
    "Product Price vs Units Sold",
    fontsize=16
)

plt.xlabel("Price (USD)", fontsize=12)
plt.ylabel("Units Sold", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=df,
    x="price_usd",
    y="units_sold",
    alpha=0.4
)

plt.title(
    "Product Price vs Units Sold",
    fontsize=16
)

plt.xlabel("Price (USD)", fontsize=12)
plt.ylabel("Units Sold", fontsize=12)

plt.tight_layout()

plt.savefig(
    "../outputs/price_vs_units_sold.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Insight

The scatter plot illustrates the relationship between product price
and units sold. The distribution of points can be used to assess
whether a clear linear relationship exists and to identify unusual
observations.

```python
total_products = len(df)

total_units_sold = df["units_sold"].sum()

average_price = df["price_usd"].mean()

average_rating = df["rating"].mean()

average_return_rate = df["return_rate"].mean()

average_discount = df["discount_percent"].mean()

print("DATASET SUMMARY")
print("=" * 40)

print(f"Total Products       : {total_products:,}")
print(f"Total Units Sold     : {total_units_sold:,}")
print(f"Average Price        : ${average_price:.2f}")
print(f"Average Rating       : {average_rating:.2f}")
print(f"Average Return Rate  : {average_return_rate:.2%}")
print(f"Average Discount     : {average_discount:.2f}%")
```

```python
top_category = (
    df.groupby("category")["units_sold"]
    .sum()
    .idxmax()
)

top_category_sales = (
    df.groupby("category")["units_sold"]
    .sum()
    .max()
)

print("Top-selling category:", top_category)
print("Total units sold:", f"{top_category_sales:,}")
```

```python
highest_return_category = (
    df.groupby("category")["return_rate"]
    .mean()
    .idxmax()
)

highest_return_rate = (
    df.groupby("category")["return_rate"]
    .mean()
    .max()
)

print("Highest-return category:", highest_return_category)
print(
    "Average return rate:",
    f"{highest_return_rate:.2%}"
)
```

```python
highest_price_category = (
    df.groupby("category")["price_usd"]
    .mean()
    .idxmax()
)

highest_average_price = (
    df.groupby("category")["price_usd"]
    .mean()
    .max()
)

print("Highest average-price category:", highest_price_category)
print(
    "Average price:",
    f"${highest_average_price:.2f}"
)
```
