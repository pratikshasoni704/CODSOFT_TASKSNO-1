```python
print("Price range:")
print(df["price_usd"].min(), "to", df["price_usd"].max())

print("\nRating range:")
print(df["rating"].min(), "to", df["rating"].max())

print("\nReturn rate range:")
print(df["return_rate"].min(), "to", df["return_rate"].max())
```

```python
cleaned_df = df.copy()

print("Original rows:", len(df))
print("Working rows:", len(cleaned_df))

before_duplicates = len(cleaned_df)

cleaned_df = cleaned_df.drop_duplicates()

after_duplicates = len(cleaned_df)

print("Duplicate rows removed:", before_duplicates - after_duplicates)
print("Rows after duplicate removal:", after_duplicates)


missing_after_duplicates = cleaned_df.isnull().sum().sum()

print("Missing values after duplicate removal:",
      missing_after_duplicates)



text_columns = [
    "category",
    "subcategory",
    "material",
    "color",
    "size",
    "season",
    "style"
]

for column in text_columns:
    cleaned_df[column] = cleaned_df[column].str.strip()
```

```python
print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
print("Duplicate product IDs:",
      cleaned_df["product_id"].duplicated().sum())
```

```python
print("Price minimum:", cleaned_df["price_usd"].min())
print("Discount range:",
      cleaned_df["discount_percent"].min(),
      "to",
      cleaned_df["discount_percent"].max())

print("Rating range:",
      cleaned_df["rating"].min(),
      "to",
      cleaned_df["rating"].max())

print("Return rate range:",
      cleaned_df["return_rate"].min(),
      "to",
      cleaned_df["return_rate"].max())
```

```python
cleaned_df.to_csv("../data/cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully.")
```
